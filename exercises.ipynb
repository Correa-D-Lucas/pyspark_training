```python
# 3. Read employees.csv with inferSchema=True
df_employees_inferschema = spark.read.options(inferSchema=True,header=True).csv("datasets/employees.csv")
df_employees_inferschema.show()
```

## Module 2

1. Select first_name column
2. Select first_name and salary
3. Filter salary > 5000
4. Filter salary between 5000 and 7000
5. Filter department_id = 10
6. Filter using AND condition
7. Filter using OR condition
8. Filter using isin()
9. Sort salaries ascending
10. Sort salaries descending

```python
# 1. Select first_name column
df_employees.select("first_name").show()
```

```python
# 2. Select first_name and salary
df_employees.select("first_name", "salary").show()
```

```python
# 3. Filter salary > 5000
df_employees.filter(df_employees.salary > 5000).show()
```

```python
# 4. Filter salary between 5000 and 7000
df_employees.filter((df_employees.salary <= 7000) & (df_employees.salary >= 5000)).show()
```

```python
# 5. Filter department_id = 10
df_employees.filter(df_employees.department_id == 10).show()
```

```python
# 6. Filter using AND condition
df_employees.filter("(salary < 7000) AND (salary > 5000)").show()
```

```python
# 7. Filter using OR condition
df_employees.filter("(salary >5000) OR (department_id ==10)").show()
```

```python
# 8. Filter using isin()
df_employees.filter(df_employees.department_id.isin([10, 20])).show()
```

```python
# 9. Sort salaries ascending
df_employees.sort(df_employees.salary.asc()).show()
```

```python
# 10. Sort salaries descending
df_employees.sort(df_employees.salary.desc()).show()
```
