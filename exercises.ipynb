## Module 4

```python
# 1. Average salary by department
from pyspark.sql.functions import avg, col, max, min, year, round, sum, count_distinct, count
df_employees.groupBy(col("department_id")) \
    .agg(
        round(avg(col("salary")),2).alias("salary_avg")
    ) \
    .orderBy(col("department_id")) \
    .show()
```

```python
# 2. Sum salary by department
df_employees.groupBy("department_id") \
    .agg(
        sum("salary").alias("total_salary")
    ) \
    .orderBy("department_id") \
    .show()
```

```python
# 3. Count employees by department
df_employees.groupBy("department_id") \
    .agg(
        count("employee_id").alias("total_employees")) \
    .orderBy("department_id") \
    .show()
```

```python
# 4. Maximum salary

df_employees.agg(max("salary").alias("max_salary")).show()
# or
df_employees.select(max("salary").alias("max_salary")).show()
```

```python
# 5. Minimum salary
df_employees.agg(min("salary").alias("min_salary")).show()
# or
df_employees.select(min("salary").alias("min_salary")).show()
```

```python
# 6. Count distinct departments
df_employees.agg(count_distinct("department_id").alias("total_departments")).show()
```

```python
# 7. Group by department and hire year
df_employees.withColumn("hire_year", year(col("hire_date"))) \
    .groupBy("department_id", "hire_year") \
        .agg(
            sum("salary").alias("total_salary")
            ) \
        .orderBy("hire_year") \
        .show()
```

```python
# 8. Order aggregation results
df_employees.groupBy("department_id") \
    .agg(
        sum("salary").alias("total_salary")
    ) \
    .orderBy(col("total_salary").desc()) \
    .show()
```

```python
# 9. Filter grouped results
df_employees.groupBy("department_id") \
    .agg(
        sum("salary").alias("total_salary")
    ) \
    .filter(col("total_salary")>15000) \
    .orderBy(col("total_salary").desc()) \
    .show()
    
```

```python
# 10. Multiple aggregations in one query
df_employees.groupBy("department_id") \
    .agg(
        round(avg("salary"),2).alias("salary_avg"),
        sum("salary").alias("total_salary"),
        count("*").alias("total_employees")
        ) \
    .orderBy("department_id") \
    .show()
```

1. row_number()
2. rank()

8. Top 3 salaries by department

```python
#1. row_number()
from pyspark.sql.functions import row_number, lag, lead, col, rank, dense_rank
from pyspark.sql.window import Window
window = (
    Window.partitionBy(col("department_id")).orderBy(col("hire_date"))
)
df_employees.withColumn("rn", row_number().over(window)).show()
```

```python
# 2. rank()

window_spec = (
    Window.partitionBy("department_id").orderBy(col("hire_date").desc())
)

df_employees.select(
    "employee_id", "salary", "department_id", "hire_date", 
    rank().over(window_spec).alias("rank")
    ) \
    .filter("rank==1") \
    .show()
```

```python
# 3. dense_rank()
window_spec = (
    Window.orderBy(col("department_id"))
)
df_employees.select("department_id", dense_rank().over(window_spec).alias("dense_rank")).show()

## IN A MASSIVE DATASET NOT SPECIFYING THE PARTITION CAN BE COSTING
```

```python
# 4. Running salary total
window_spec = (
    Window.partitionBy("department_id").orderBy("hire_date")
)

df_employees.select("department_id", sum("salary").over(window_spec).alias("running_salary")).show()
```

```python
# 5. Moving average
window_spec = (
    Window.partitionBy("department_id")
          .orderBy("hire_date")
          .rowsBetween(-2, 0)
)

df_employees.select(
    "department_id",
    "hire_date",
    "salary",
    avg("salary").over(window_spec).alias("moving_avg_salary")
).show()
```

```python
# 6. lag()
window_spec = (
    Window.partitionBy("department_id").orderBy("hire_date")
)
df_employees.select("department_id", "hire_date", lag("hire_date").over(window_spec).alias("lag_date")).show()
```

```python
# 7. lead()
window_spec = (
    Window.partitionBy("department_id").orderBy("hire_date")
)


df_employees.select("department_id", "hire_date", lead("hire_date").over(window_spec).alias("lag_date")).show()
```

```python
# 8. Top 2 salaries by department

window_spec = (
    Window.partitionBy("department_id").orderBy(col("salary").desc())
)

df_employees.select("department_id", "first_name", "salary", row_number().over(window_spec).alias("rn")).filter("rn IN (1,2)").show()
```

```python
# 9. First employee by hire_date
window_spec = (
    Window.partitionBy("department_id").orderBy(col("hire_date"))
)

df_employees.select("department_id", "first_name","hire_date", "salary", row_number().over(window_spec).alias("rn")).filter("rn = 1").show()
```

1. Convert string to date

```python
# 1. Convert string to date

data = [
    (1, "2020-01-01"),
    (2, "2022-01-01"),
    (3, "2025-01-01"),
    (4, "2026-01-01"),
    (5, "2027-01-01")
]

column_names = [
    "id", "date"
]

df = spark.createDataFrame(data, column_names)
df.printSchema()
df.select("id", col("date").cast("date").alias("date")).printSchema()
```

```python
# 2. Extract year
df = df.select("id", col("date").cast("date").alias("date"))
df.withColumn("year", year(col("date"))).show()

```

```python
#3. Extract month
df.withColumn("month", month(col("date"))).show()
```

```python
# 4. Extract day
from pyspark.sql.functions import day
df.withColumn("day", day(col("date"))).show()
```

```python
# 5. Calculate date difference
from pyspark.sql.functions import lead, datediff, to_date, row_number, dateadd

data = [
    (101, "Lucas", "2024-01-15"),
    (101, "Lucas", "2024-03-10"),
    (101, "Lucas", "2024-06-20"),
    (102, "Maria", "2024-02-01"),
    (102, "Maria", "2024-05-15"),
    (102, "Maria", "2024-09-01"),
    (103, "John", "2024-01-05"),
    (103, "John", "2024-04-18")
]

column_names = [
    "employee_id",
    "employee_name",
    "event_date"
]

df1 = spark.createDataFrame(data, column_names)

df1 = df1.withColumn(
    "event_date",
    to_date("event_date")
)

window_spec = (
    Window.partitionBy("employee_name")
          .orderBy("event_date")
)

(
    df1
    .withColumn(
        "next_event_date",
        lead("event_date").over(window_spec)
    )
    .withColumn(
        "date_diff",
        datediff("next_event_date", "event_date")
    )
    .orderBy("employee_id")
    .show()
)
```

```python
# 6. Add days

window_spec = (
    Window.partitionBy("employee_name").orderBy(col("event_date").desc())
)

df1.withColumn(
    "event_order", row_number().over(window_spec)
) \
.filter("event_order = 1") \
.withColumn("next_event", dateadd("event_date", 30)) \
.show()

```

```python
# 7. Add months

from pyspark.sql.functions import add_months

window_spec = (
    Window.partitionBy("employee_name").orderBy(col("event_date").desc())
)

df1.withColumn(
    "event_order", row_number().over(window_spec)
) \
.filter("event_order = 1") \
.withColumn("next_event", add_months("event_date", 1)) \
.show()
```

```python
# 8. Filter date range

df1.select("*").filter("event_date >= '2024-02-01' and event_date <= '2024-05-15'").show()
```

```python
# 9. Create timestamp
from pyspark.sql.functions import to_timestamp

df1.withColumn("event_timestamp", to_timestamp("event_date")).show()
```

```python
# 10. Work with current_date()
from pyspark.sql.functions import current_date
df1.withColumn("current_date", current_date()).withColumn("days_until_today", datediff("current_date", "event_date")).show()
```

## Module 8

1. Detect nulls
2. Drop nulls
3. Fill nulls
4. Replace null salary with average
5. Remove duplicates
6. Trim spaces
7. Replace values
8. Regex replace
9. Standardize text
10. Detect inconsistent rows

```python
from pyspark.sql import functions as T
data = [
    (1,"lucas", "data", 10000),
    (1,"lucas", "data", 10000),
    (2,"mateus", "operacao", 7000),
    (3,"pedro", "saude", 7000),
    (4,"jessica", "saude", None),
    (5,"sabrina", "data", 15000),
    (6,"bruno", "data", 9000),
    (7,"laryssa", "operacao", 5000),
    (8,"joilson", "saude", 7000),
    (9,"gisele", "operacao", 7500)
]

column_names = ["employee_id", "employee_name", "department_name", "salary"]



df = spark.createDataFrame(data=data, schema=column_names)
```

```python
# 1. Detect nulls
from functools import reduce

condition = reduce(
    lambda x, y: x | y,
    [T.col(c).isNull() for c in df.columns]

)

df.filter(condition).show()
```

```python
# 2. Drop nulls
df.dropna().show()
```

```python
# 3. Fill nulls
# 4. Replace null salary with average
from pyspark.sql import functions as F
from pyspark.sql.window import Window

window_spec = Window.partitionBy("department_name")

df_filled = (
    df.withColumn(
        "salary",
        F.when(
            F.col("salary").isNull(),
            F.avg("salary").over(window_spec)
        ).otherwise(F.col("salary"))
    )
)

df_filled.show()
```

```python
# 5. Remove duplicates
df.dropDuplicates().show()
```

```python
# 6. Trim spaces
df.select("*", T.trim(T.col("employee_name")).alias("employee_name_trimmed")).show()
```

```python
# 7. Replace values
df.fillna({"salary":0}).show()
```

```python
# 8. Regex replace
from pyspark.sql.functions import regexp_replace
from pyspark.sql import functions as T
df.select("*", T.concat(T.col("employee_name"), T.lit("_testing")).alias("new")).withColumn("new", T.regexp_replace("new", r"_testing", "@")).show()
```

```python
# 9. Standardize text
df.withColumn("standardize_name", T.lower(T.trim(T.col("employee_name")))).show()
```

```python
# 10. Detect inconsistent rows
df.filter(T.col("salary").isNull()).show()
```

```python
# 7. INNER JOIN
df_employees = spark.read.parquet("datasets/employees_write_partitioned.parquet")
df_departments = spark.read.options(header=True, inferSchema=True).csv("datasets/departments.csv")

df_employees.createOrReplaceTempView("employees")
df_departments.createOrReplaceTempView("departments")
spark.sql("""
    SELECT d.department_id, d.department_name, e.first_name, e.salary
    FROM departments as d
    JOIN employees as e
    ON d.department_id = e.department_id
    """).show()
```

```python
# 10. Subquery
spark.sql("""SELECT first_name, salary 
FROM employees
WHERE salary > (SELECT AVG(salary) FROM employees)""").show()
```

```python
# 11. CTE

spark.sql("""
    WITH data_dept AS (
          SELECT *
          FROM departments
          WHERE lower(department_name) LIKE "data"
        )
    SELECT e.first_name, e.salary
    FROM employees as e
    INNER JOIN data_dept as dp ON e.department_id = dp.department_id
          """).show()
```
